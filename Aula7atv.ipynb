```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns


```

```python
#PARTE1
```

```python
arquivo = "Tabela_13_Meio_de_transporte.xlsx"

df = pd.read_excel(
    arquivo,
    sheet_name="Exemplo gráfico Brasil"
)

df
```

```python

df.columns = [
    'Meio de Transporte',
    'Branca',
    'Preta ou Parda',
    'Indígena'
]


grupos = ['Branca', 'Preta ou Parda', 'Indígena']

for col in grupos:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace('%', '', regex=False)
        .str.replace(',', '.', regex=False)
        .str.strip()
    )

    df[col] = pd.to_numeric(df[col], errors='coerce')

df
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

cores = [
    '#ff9999',
    '#66b3ff',
    '#99ff99',
    '#ffcc99',
    '#c2c2f0',
    '#ffb3e6'
]

grupos = ['Branca', 'Preta ou Parda', 'Indígena']

for i, col in enumerate(grupos):


    dados = df[[col, 'Meio de Transporte']].dropna()

    axes[i].pie(
        dados[col],
        labels=dados['Meio de Transporte'],
        autopct='%1.3f%%',
        startangle=180,
        colors=cores[:len(dados)],
        wedgeprops={
            'edgecolor': 'white',
            'linewidth': 1.2
        }
    )

    axes[i].set_title(
        f'Cor/Raça: {col}',
        fontsize=12,
        fontweight='bold'
    )

plt.suptitle(
    'Meios de transporte utilizados pelas mulheres por cor ou raça - Brasil',
    fontsize=14,
    fontweight='bold'
)

plt.tight_layout()

plt.savefig(
    'grafico_transporte.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
df2 = pd.read_excel(
    arquivo,
    sheet_name="BR GR UF MU",
    header=None
)

df2.head()
```

```python
df2 = df2.iloc[3:].copy()

df2.columns = [
    'Local',
    'Branca_A_pe',
    'Branca_Bicicleta',
    'Branca_Motocicleta',
    'Branca_Carro',
    'Branca_Transporte_Coletivo',
    'Branca_Outros',
    
    'Preta_Parda_A_pe',
    'Preta_Parda_Bicicleta',
    'Preta_Parda_Motocicleta',
    'Preta_Parda_Carro',
    'Preta_Parda_Transporte_Coletivo',
    'Preta_Parda_Outros',
    
    'Indigena_A_pe',
    'Indigena_Bicicleta',
    'Indigena_Motocicleta',
    'Indigena_Carro',
    'Indigena_Transporte_Coletivo',
    'Indigena_Outros'
]

for coluna in df2.columns[1:]:
    df2[coluna] = pd.to_numeric(
        df2[coluna],
        errors='coerce'
    )

df2.head()
```

```python
estados = [
    'Rondônia',
    'Acre',
    'Amazonas',
    'Roraima',
    'Pará',
    'Amapá',
    'Tocantins',
    'Maranhão',
    'Piauí',
    'Ceará',
    'Rio Grande do Norte',
    'Paraíba',
    'Pernambuco',
    'Alagoas',
    'Sergipe',
    'Bahia',
    'Minas Gerais',
    'Espírito Santo',
    'Rio de Janeiro',
    'São Paulo',
    'Paraná',
    'Santa Catarina',
    'Rio Grande do Sul',
    'Mato Grosso do Sul',
    'Mato Grosso',
    'Goiás',
    'Distrito Federal'
]

df_estados = df2[df2['Local'].isin(estados)].copy()

df_estados[['Local']]
```

```python
colunas_coletivo = [
    'Branca_Transporte_Coletivo',
    'Preta_Parda_Transporte_Coletivo',
    'Indigena_Transporte_Coletivo'
]

df_estados['Media_Transporte_Coletivo'] = (
    df_estados[colunas_coletivo]
    .mean(axis=1)
)

maior_coletivo = df_estados.loc[
    df_estados['Media_Transporte_Coletivo'].idxmax()
]

print(
    "Estado com maior percentual médio de mulheres utilizando transporte coletivo:"
)

print(maior_coletivo[
    [
        'Local',
        'Branca_Transporte_Coletivo',
        'Preta_Parda_Transporte_Coletivo',
        'Indigena_Transporte_Coletivo',
        'Media_Transporte_Coletivo'
    ]
])
```

```python
menor_coletivo = df_estados.loc[
    df_estados['Media_Transporte_Coletivo'].idxmin()
]

print(
    "Estado com menor percentual médio de mulheres utilizando transporte coletivo:"
)

print(menor_coletivo[
    [
        'Local',
        'Branca_Transporte_Coletivo',
        'Preta_Parda_Transporte_Coletivo',
        'Indigena_Transporte_Coletivo',
        'Media_Transporte_Coletivo'
    ]
])
```

```python
# remove estados e regiões para trabalhar somente com municípios
regioes = [
    'Brasil',
    'Norte',
    'Nordeste',
    'Sudeste',
    'Sul',
    'Centro-oeste'
]

df_cidades = df2[
    ~df2['Local'].isin(estados + regioes)
].copy()

colunas_bicicleta = [
    'Branca_Bicicleta',
    'Preta_Parda_Bicicleta',
    'Indigena_Bicicleta'
]

df_cidades['Media_Bicicleta'] = (
    df_cidades[colunas_bicicleta]
    .mean(axis=1)
)

maior_bicicleta = df_cidades.loc[
    df_cidades['Media_Bicicleta'].idxmax()
]

print("Cidade com maior percentual médio de mulheres utilizando bicicleta:")

print(
    maior_bicicleta[
        [
            'Local',
            'Branca_Bicicleta',
            'Preta_Parda_Bicicleta',
            'Indigena_Bicicleta',
            'Media_Bicicleta'
        ]
    ]
)
```

```python
colunas_carro = [
    'Branca_Carro',
    'Preta_Parda_Carro',
    'Indigena_Carro'
]

df_cidades['Media_Carro'] = (
    df_cidades[colunas_carro].mean(axis=1)
)

maior_carro = df_cidades.loc[
    df_cidades['Media_Carro'].idxmax()
]

print("Cidade com maior percentual médio de mulheres utilizando carro:")

print(
    maior_carro[
        [
            'Local',
            'Branca_Carro',
            'Preta_Parda_Carro',
            'Indigena_Carro',
            'Media_Carro'
        ]
    ]
)
```
